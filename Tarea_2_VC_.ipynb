**3.1.Cargue los datos iris en un data frame (pandas) e imprima

**3.3.Obtenga el número de muestras faltantes o Nan.**

```python
nulos_sepal_length = df['sepal_length'].isnull().sum()
print(nulos_sepal_length," Registros nulos en sepal_length")


nulos_sepal_width = df['sepal_width'].isnull().sum()
print(nulos_sepal_width," Registros nulos en sepal_width")

nulos_petal_length = df['petal_length'].isnull().sum()
print(nulos_petal_length," Registros nulos en petal_length")

nulos_petal_width = df['petal_width'].isnull().sum()
print(nulos_petal_width," Registros nulos en petal_width")
```

**3.4.Cree un arreglo 2-D de tamaño 5x5 con unos en la diagonal y ceros en el resto. Convierta el arreglo NumPy
a una matriz dispersa de ScyPy en formato CRS. Nota: una matriz se considera dispersa cuando el
porcentaje de ceros es mayor a 0.5.**

```python
dosD = np.eye(5)
matriz_dispersa = csr_matrix(dosD)

# Mostrar la matriz dispersa
print("Matriz dispersa en formato CRS:")
print(matriz_dispersa)
```

**3.9.Cree una gráfica de barras que muestre la media, mínimo y máximo de todos los datos.**

```python
# Suponiendo que media, min_val y max_val ya están definidos
resumen = pd.DataFrame({'Media': media, 'Mínimo': min, 'Máximo': max})

plt.figure(figsize=(10, 6))
resumen.plot(kind='bar', figsize=(10, 6), alpha=0.7)




plt.xlabel("Clases")
plt.ylabel("Valores")
plt.title("Media, Mínimo y Máximo de las clases")
plt.xticks(rotation=45)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
```

**3.10. Muestre la frecuencia de las tres especies como una gráfica de pastel.**

```python
clases = df['class'].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(clases, labels=clases.index, autopct='%1.1f%%', startangle=140)

plt.title("Distribución de Especies")

plt.show()
```

**3.11. Cree una gráfica que muestre la relación entre la longitud y ancho del sépalo de las tres especies
conjuntamente.**

```python

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x=df["sepal_length"], y=df["sepal_width"], hue="class")


plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Ancho del sépalo (cm)")
plt.title("Longitud vs el ancho del sépalo en las especies")
plt.legend(title="Especies")

plt.show()
```

**3.12. Obtenga los histogramas de las variables SepalLength, SepalWidth, PetalLength y PetalWidth.**

```python

columns = df.columns[:-1]


plt.figure(figsize=(12, 8))
for i, col in enumerate(columns, 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[col], bins=15, kde=True)
    plt.title(f"Histograma de {col}")
    plt.xlabel(col)
    plt.ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

**3.14. Cree una gráfica usando joinplot de seaborn para mostrar la dispersión entre la longitud y ancho
del sépalo y las distribuciones de estas dos variables.**

```python

# Crear la gráfica de dispersión conjunta (jointplot)
sns.jointplot(
    data=df,
    x=df["sepal_width"],
    y=df['petal_length'],
    hue="class",
    kind="scatter",
    height=8
)
plt.legend(title="Especies")
plt.show()
```
